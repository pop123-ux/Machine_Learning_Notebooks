```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
```

```python
# Term Frequency-Inverse Document Frequency Vectorizing

from sklearn.feature_extraction.text import TfidfVectorizer

X_train = train_df['text']
y_train = train_df['label']

X_test = test_df['text']

vectorizer = TfidfVectorizer(sublinear_tf=True, lowercase=False, stop_words='english')

X_train_trans = vectorizer.fit_transform(X_train)
X_test_trans = vectorizer.transform(X_test)
```

```python
# Light Gradient Boosting Classifier initialization + training

from lightgbm import LGBMClassifier

model = LGBMClassifier(objective='multiclass', n_estimators=300, boosting_type='dart', n_jobs=-1)

model.fit(X_train_trans, y_train)
```

```python
y_preds = model.predict(X_test_trans)
```

```python
pd.DataFrame(y_preds).value_counts()
```

```python
# Verification

len(y_preds) == len(test_df)
```

```python
# Submission DataFrame building + downloading

S = pd.DataFrame({
    'SampleID': test_df['SampleID'],
    'label': y_preds,
})

S.head(1)
```
