```python
"""Now let's get back to our EDA (Exploratory Data Analysis)"""
np.unique(dataset.target)
```

```python
"""Let's create a dataframe now"""
df = pd.DataFrame(dataset.data, columns=dataset.feature_names)
df.head()
```

```python
df.describe()
```

```python
"""Let's now get started with the Machine Learning part"""
X = df
y = dataset.target
```

```python
"""Z-score scaling --> mean=0; var=1"""
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
"""train_test_split on the curated digits dataset"""
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
"""Training"""
from catboost import CatBoostClassifier
from catboost import Pool

model = CatBoostClassifier(depth=4, random_strength=0.4, verbose=200, random_seed=42)
model.fit(X_train, y_train)
print('\n')
print(f"Model Accuracy on Val + Train splits: {round(model.score(X_test, y_test), 3)}, {model.score(X_train, y_train)}")
```

```python
eval_data = Pool(X_test, y_test)

metrics_result = model.eval_metrics(data=eval_data, metrics=['Accuracy'])
```

```python
"""Let's now plot the Accuracy"""
plt.plot(metrics_result['Accuracy'], c='r')
plt.title("Accuracy on Val split")
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()
```

```python
"""Let's now do PCA and analyse some methods of it's class"""
from sklearn.decomposition import PCA

pca = PCA(0.95) # Retain 95% of useful features and then create new dimensions

X_pca = pca.fit_transform(X)
print(f"Original DataFrame shape vs newly decomposed DataFrame shape: {X.shape}, {X_pca.shape}")
```

```python
pca.explained_variance_ratio_ # Tells us how much each column captures actual useful information (in percentages)
```

```python
pca.n_components_ == X_pca.shape[1] == 29  # How many components do we have = nr. of columns
```

```python
"""train_test_split again, now with PCA decomposition"""
X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=42)
```

```python
"""Training"""
model = CatBoostClassifier(depth=4, random_strength=0.4, verbose=200, random_seed=42)
model.fit(X_train_pca, y_train)
print('\n')
print(f"Model Accuracy on Val + Train splits: {round(model.score(X_test_pca, y_test), 3)}, {model.score(X_train_pca, y_train)}")
```

```python
"""Let's see what happens if we want 2 components - It will identify the most important 2 features of the dataset"""
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
X_pca.shape
```

```python
pca.n_components_ == X_pca.shape[1] == 2  # How many components do we have = nr. of columns
```

```python
X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=42)

"""Training Loop"""
model = CatBoostClassifier(depth=4, iterations=1000, random_strength=0.4, verbose=200, random_seed=42)
model.fit(X_train_pca, y_train)
print('\n')
print(f"Model Accuracy on Val + Train splits (with only 2 components!): {round(model.score(X_test_pca, y_test), 3)}, {model.score(X_train_pca, y_train)}")
```

## As we can see, due to reduction of 27 components we can observe the accuracy differences between the 2 test runs: **97.5% vs. 66.4%**

## By eliminating 25 dimensions we sacrifice approximately 31% Accuracy!

```python
"""Evaluate & Plot"""
eval_data_train = Pool(X_train_pca, y_train)
eval_data_test = Pool(X_test_pca, y_test)

metrics_result_train = model.eval_metrics(eval_data_train, metrics=['Accuracy', 'F1'])
metrics_result_test = model.eval_metrics(eval_data_test, metrics=['Accuracy', 'F1'])
```

```python
fig, (ax1, ax2) = plt.subplots(figsize=(14,6), nrows=1, ncols=2)

ax1.plot(metrics_result_test['Accuracy'], c='r')
ax1.set_title("Accuracy on Test split (w/ 2 components)")
ax1.set_ylabel('Accuracy')
ax1.grid(True)

ax2.plot(metrics_result_train['Accuracy'], c='b')
ax2.set_title('Accuracy on Train split (w/ 2 components)')
ax2.set_ylabel('Accuracy')
ax2.grid(True)

plt.tight_layout()
plt.show()
```
